### Load Data

```python
# load data
from src.data_loader import load_data

df = load_data()
```

```python
import pandas as pd
```

### Handle Data  Types

```python
df.dtypes
```

```python
df["Total Charges"] = pd.to_numeric(df["Total Charges"], errors='coerce')
```

### Handle Missing Values

```python
df.isna().sum()
```

### Removing Unnecessary Columns

```python
DROP_COLUMNS = [ "CustomerID",
                "Count",
                "Country",
                "State",
                "City",
                "Zip Code",
                "Lat Long",
                "Latitude",
                "Longitude",
                "Churn Score",
                "CLTV",
                "Churn Reason"
]                                
                
```

```python
df = df.drop(columns=DROP_COLUMNS, errors="ignore")
```

```python
# remove missing entries rows
df = df.dropna(subset=["Total Charges"]).copy()
```

### Encode Target Variable

```python
# Converting the target to binary

df["Encoded Churn Label"] = df["Churn Label"].map({"Yes": 1, "No": 0})
```

### Feature Seperation

```python
# Seperate Features
X = df.drop(columns=["Encoded Churn Label", "Churn Label", "Churn Value"])
y = df["Encoded Churn Label"]
numerical_features = X.select_dtypes(include=["int64", "float64"]).columns
print("Numerical Features:", numerical_features)

categorical_features = X.select_dtypes(include=["object"]).columns
print("Categorical Features:", categorical_features)
```

### Saving Processed Dataset

```python
# Save the cleaned dataset

df.to_csv(PROJECT_ROOT / "data" / "processed" / "telco_processed.csv", index=False)
```
